```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def plot_bernoulli_trials(num_trials, probability_of_success):
    # Generate Bernoulli trials
    outcomes = np.random.binomial(n=1, p=probability_of_success, size=num_trials)
    
    # Plotting 
    plt.figure(figsize=(8, 4))

    # Plotting the outcomes
    for i, outcome in enumerate(outcomes):
        if outcome == 1:
            square = plt.Rectangle((i, 0), 1, 1, fill=False, edgecolor='black')
            plt.gca().add_patch(square)
            circle = plt.Circle((i + 0.5, 0.5), 0.2, fill=False, edgecolor='blue')
            plt.gca().add_patch(circle)
        else:
            square = plt.Rectangle((i, 0), 1, 1, fill=False, edgecolor='black')
            plt.gca().add_patch(square)

    plt.xlim(0, num_trials)
    plt.ylim(0, 1)
    plt.xlabel('Trial Number')
    plt.title('Bernoulli Trials')
    plt.gca().axes.get_yaxis().set_visible(False)
    plt.gca().set_aspect('equal', adjustable='box')  # Ensure aspect ratio is preserved
    plt.show()

# Define interactive widgets
num_trials_slider = widgets.IntSlider(value=30, min=1, max=100, step=1, description='Number of Trials:')
prob_success_slider = widgets.FloatSlider(value=0.3, min=0, max=1, step=0.01, description='Probability of Success:')

# Create interactive output
interactive_plot = widgets.interactive_output(plot_bernoulli_trials, 
                                              {'num_trials': num_trials_slider,
                                               'probability_of_success': prob_success_slider})

# Display widgets and output
display(num_trials_slider, prob_success_slider, interactive_plot)
```
