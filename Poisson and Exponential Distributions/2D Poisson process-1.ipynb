```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import ipywidgets as widgets
from IPython.display import display

def plot_poisson_process(area_width, lambda_param):
    # Generate points in the 2D Poisson process
    num_points = np.random.poisson(lambda_param)
    points_x = np.random.uniform(0, area_width, num_points)
    points_y = np.random.uniform(0, area_width, num_points)

    # Plotting
    plt.figure(figsize=(8, 8))

    # Plotting the circles representing event occurrences
    for x, y in zip(points_x, points_y):
        circle = patches.Circle((x, y), radius=0.2, fill=False, edgecolor='b')
        plt.gca().add_patch(circle)

    plt.xlim(0, area_width)
    plt.ylim(0, area_width)

    # Increase font size
    plt.xlabel('X', fontsize=24)
    plt.ylabel('Y', fontsize=24)
    plt.title('2D Spatial Poisson Process', fontsize=16)
    
    # Increase tick label size
    plt.xticks(fontsize=24)
    plt.yticks(fontsize=24)

    plt.gca().set_aspect('equal', adjustable='box')  # Ensure aspect ratio is preserved
    plt.show()

# Define interactive widgets
area_width_slider = widgets.FloatSlider(value=10, min=0.1, max=20, step=0.1, description='Area Width:')
lambda_slider = widgets.FloatSlider(value=10, min=0.1, max=20, step=0.1, description='Lambda:')

# Create interactive output
interactive_plot = widgets.interactive_output(plot_poisson_process, 
                                              {'area_width': area_width_slider,
                                               'lambda_param': lambda_slider})

# Display widgets and output
display(area_width_slider, lambda_slider, interactive_plot)
```
