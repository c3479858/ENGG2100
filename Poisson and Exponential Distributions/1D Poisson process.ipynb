```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import ipywidgets as widgets
from IPython.display import display

def plot_poisson_process(lambda_param, interval_length):
    # Generate points in the Poisson process
    num_points = np.random.poisson(lambda_param)
    points = np.sort(np.random.uniform(0, interval_length, num_points))
     
    # Plotting
    plt.figure(figsize=(10, 1))

    # Plotting the circles
    for point in points:
        circle = patches.Circle((point, 0), radius=0.2, fill=False, edgecolor='b')
        plt.gca().add_patch(circle)

    plt.xlim(0, interval_length)
    plt.ylim(-0.3, 0.3)
    plt.xlabel('Position')
    plt.title('Poisson Process')
    plt.gca().axes.get_yaxis().set_visible(False)
    plt.gca().set_aspect('equal', adjustable='box')  # Ensure aspect ratio is preserved
    plt.show()

# Define interactive widgets
lambda_slider = widgets.FloatSlider(value=15, min=0.1, max=50, step=0.1, description='λ:')
interval_length_slider = widgets.FloatSlider(value=30, min=1, max=100, step=1, description='Interval Length:')

# Create interactive output
interactive_plot = widgets.interactive_output(plot_poisson_process, 
                                              {'lambda_param': lambda_slider,
                                               'interval_length': interval_length_slider})

# Display widgets and output
display(lambda_slider, interval_length_slider, interactive_plot)
```
