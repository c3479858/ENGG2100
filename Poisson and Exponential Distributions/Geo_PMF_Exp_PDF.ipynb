```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import expon, geom
import ipywidgets as widgets
from IPython.display import display

def plot_distributions(p_param_geom, n_param_geom, x_limit):
    # Parameters for the distributions
    lambda_param_exp = p_param_geom  # Rate parameter for the exponential distribution

    # Generate x values for plotting
    x_geom = np.arange(1, n_param_geom + 1)  # Geometric distribution: integers from 1 to n
    x_exp = np.arange(0, x_limit + 1, 1)  # Exponential distribution: integers from 0 to x_limit

    # Calculate the PMF for the geometric distribution
    pmf_geom = geom.pmf(x_geom, p=p_param_geom)

    # Calculate the PDF for the exponential distribution
    pdf_exp = expon.pdf(x_exp, scale=1/lambda_param_exp)

    # Plot the PMF and PDF
    plt.figure(figsize=(10, 6))
    plt.bar(x_geom, pmf_geom, color='green', alpha=0.7, label=f'Geometric (p={p_param_geom}, n={n_param_geom})')
    plt.plot(x_exp, pdf_exp, color='blue', label=f'Exponential (λ={lambda_param_exp})')
    plt.title('Comparison of PMF and PDF: Exponential vs Geometric')
    plt.xlabel('Value')
    plt.ylabel('Probability')
    plt.legend()
    plt.grid(True)

    # Set x-axis tick locations with an increment of 1
    plt.xticks(np.arange(min(x_exp), max(x_exp)+1, 1))

    # Set x-axis limit
    plt.xlim(0, x_limit)

    plt.show()

# Define interactive widgets
p_slider = widgets.FloatSlider(value=0.01, min=0.001, max=1.0, step=0.001, description='p:')
n_slider = widgets.IntSlider(value=500, min=1, max=1000, step=1, description='n:')
x_limit_slider = widgets.IntSlider(value=20, min=1, max=1500, step=1, description='x limit:')

# Create interactive output
interactive_plot = widgets.interactive_output(plot_distributions, 
                                              {'p_param_geom': p_slider,
                                               'n_param_geom': n_slider,
                                               'x_limit': x_limit_slider})

# Display widgets and output
display(p_slider, n_slider, x_limit_slider, interactive_plot)
```
