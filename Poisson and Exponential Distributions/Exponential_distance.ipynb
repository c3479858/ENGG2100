```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def plot_spatial_poisson(num_points, interval_length):
    # Average rate of points per unit length
    lambda_param = num_points / interval_length

    # Generate points in the spatial Poisson process
    points = np.sort(np.random.uniform(0, interval_length, num_points))

    # Calculate distances between adjacent points
    distances = np.diff(points)

    # Plot histogram of distances
    plt.hist(distances, bins=30, density=True, alpha=0.6, color='b', label='Empirical distribution')

    # Plot the theoretical exponential distribution
    x = np.linspace(0, max(distances), 100)
    exponential_dist = lambda_param * np.exp(-lambda_param * x)
    plt.plot(x, exponential_dist, 'r', label='Exponential distribution')

    plt.xlabel('Distance')
    plt.ylabel('Probability density')
    plt.title('Distances between Adjacent Points in Spatial Poisson Process')
    plt.legend()
    plt.grid(True)
    plt.show()

# Define interactive widgets
num_points_slider = widgets.IntSlider(value=1000, min=10, max=5000, step=10, description='Number of Points:')
interval_length_slider = widgets.FloatSlider(value=10, min=1, max=100, step=1, description='Interval Length:')

# Create interactive output
interactive_plot = widgets.interactive_output(plot_spatial_poisson, 
                                              {'num_points': num_points_slider,
                                               'interval_length': interval_length_slider})

# Display widgets and output
display(num_points_slider, interval_length_slider, interactive_plot)
```
