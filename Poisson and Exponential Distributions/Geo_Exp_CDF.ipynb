```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import geom
import ipywidgets as widgets
from IPython.display import display

def plot_cdfs(p_param_geom, n_param_geom, x_limit):
    # Generate x values for plotting
    x = np.linspace(0, x_limit, 1000)

    # Calculate lambda based on p
    lambda_param_exp = p_param_geom

    # Calculate the CDFs
    cdf_exp = expon.cdf(x, scale=1/lambda_param_exp)
    cdf_geom = geom.cdf(x, p=p_param_geom)

    # Plot the CDFs
    plt.figure(figsize=(10, 6))
    plt.plot(x, cdf_exp, color='blue', label=f'Exponential (λ = {lambda_param_exp})')
    plt.plot(x, cdf_geom, color='green', label=f'Geometric (p = {p_param_geom}, n = {n_param_geom})')
    plt.title('Comparison of CDFs: Exponential vs Geometric')
    plt.xlabel('Value')
    plt.ylabel('Cumulative Probability')
    plt.legend()
    plt.grid(True)
    plt.xlim(0, x_limit)
    plt.show()

# Define interactive widgets
p_slider = widgets.FloatSlider(value=0.01, min=0.001, max=1.0, step=0.001, description='p:')
n_slider = widgets.IntSlider(value=200, min=1, max=1000, step=1, description='n:')
x_limit_slider = widgets.IntSlider(value=200, min=1, max=500, step=1, description='x limit:')

# Create interactive output
interactive_plot = widgets.interactive_output(plot_cdfs, 
                                              {'p_param_geom': p_slider,
                                               'n_param_geom': n_slider,
                                               'x_limit': x_limit_slider})

# Display widgets and output
display(p_slider, n_slider, x_limit_slider, interactive_plot)
```
